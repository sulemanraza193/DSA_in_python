```python
class Node():
    def __init__(self, value):
        self.value = value
        self.next = None
        self.pre = None

class DoublyLL():
    def __init__(self):
        self.head = None

    def insertAtEnd(self, value):
        temp = Node(value)

        if(self.head == None):
            self.head = temp
            return

        t = self.head
        while(t.next != None):
            t = t.next

        t.next = temp
        temp.pre = t
    def insertAtBegin(self, value):
        temp = Node(value)

        if(self.head == None):
            self.head = temp
            return

        temp.next = self.head
        self.head.pre = temp
        self.head = temp

    def insertAtPos(self, value, pos):
        temp = Node(value)

        if(pos == 1):
            self.insertAtBegin(value)
            return

        t = self.head
        for i in range(1, pos-1):
            t = t.next

        temp.next = t.next
        temp.pre = t
        t.next = temp
        if(temp.next != None):
            temp.next.pre = temp
    def deleteAtPos(self, pos):
        if(pos == 1):
            self.head = self.head.next
            if(self.head != None):
                self.head.pre = None
            return

        t = self.head
        for i in range(1, pos-1):
            t = t.next

        t.next = t.next.next
        if(t.next != None):
            t.next.pre = t

    def printDLL(self):
        t = self.head
        while(t != None):
            print(t.value, end="<-->")
            t = t.next
        print()
  
obj = DoublyLL()
obj.insertAtEnd(10)
obj.insertAtEnd(20)
obj.insertAtEnd(30)
obj.insertAtEnd(40)
obj.insertAtEnd(50)
obj.printDLL()
obj.insertAtBegin(5)
obj.insertAtPos(25, 4)


obj.printDLL()

        
```
