```python
class Node:
    def __init__(self, info, next=None):
        self.data = info
        self.next = next

class SinglyCircularList():
    def __init__(self):
        self.head = None
        self.last = None

    def insertAtEnd(self, value):
        temp = Node(value)

        if self.head is not None:
            t1 = self.head
            while t1.next is not self.head:
                t1 = t1.next
            t1.next = temp
            temp.next = self.head  # Essential to maintain circular structure
            self.last = temp       # Keep self.last updated
        else:
            self.head = temp
            self.last = temp
            temp.next = self.head  # Single node points to itself

    def insertAtBeginning(self, value):
        temp = Node(value)

        if self.head is not None:
            temp.next = self.head
            self.head = temp
            self.last.next = self.head  # Update last node's next to new head
        else:
            self.head = temp
            self.last = temp
            temp.next = self.head  # Single node points to itself

    def insertAfter(self, value, afterValue):
        temp = Node(value)

        if self.head is None:
            print("List is empty")
            return

        t1 = self.head
        while True:
            if t1.data == afterValue:
                temp.next = t1.next
                t1.next = temp
                if t1 == self.last:  # Update last if we inserted after the last node
                    self.last = temp
                return
            t1 = t1.next
            if t1 == self.head:  # If we have looped back to head, value not found
                print(f"Value {afterValue} not found in the list.")
                return
    def deleteNode(self, value):
        if self.head is None:
            print("List is empty")
            return

        t1 = self.head
        prev = None

        while True:
            if t1.data == value:
                if prev is not None:
                    prev.next = t1.next
                    if t1 == self.last:  # Update last if we are deleting the last node
                        self.last = prev
                else:
                    # Deleting the head node
                    if t1.next == self.head:  # Only one node in the list
                        self.head = None
                        self.last = None
                    else:
                        self.head = t1.next
                        self.last.next = self.head  # Update last node's next to new head
                return
            prev = t1
            t1 = t1.next
            if t1 == self.head:  # If we have looped back to head, value not found
                print(f"Value {value} not found in the list.")
                return

    def printLL(self):
            t1 = self.head
            if t1 is None:
                print("List is empty")
                return
            
            while t1 is not self.last:
                print(t1.data)
                t1 = t1.next
            print(t1.data)  # Print the last node's data

obj = SinglyCircularList()
obj.insertAtEnd(10)
obj.insertAtEnd(20)
obj.insertAtEnd(30)
obj.insertAtEnd(40)
obj.insertAtBeginning(5)
obj.insertAfter(25, 20)


obj.printLL()
```

```python
class Node():
    def __init__(self, value):
        self.value = value
        self.next = None
        self.pre = None

class DoublyLL():
    def __init__(self):
        self.head = None

    def insertAtEnd(self, value):
        temp = Node(value)

        if(self.head == None):
            self.head = temp
            return

        t = self.head
        while(t.next != None):
            t = t.next

        t.next = temp
        temp.pre = t
    def insertAtBegin(self, value):
        temp = Node(value)

        if(self.head == None):
            self.head = temp
            return

        temp.next = self.head
        self.head.pre = temp
        self.head = temp

    def insertAtPos(self, value, pos):
        temp = Node(value)

        if(pos == 1):
            self.insertAtBegin(value)
            return

        t = self.head
        for i in range(1, pos-1):
            t = t.next

        temp.next = t.next
        temp.pre = t
        t.next = temp
        if(temp.next != None):
            temp.next.pre = temp
    def deleteAtPos(self, pos):
        if(pos == 1):
            self.head = self.head.next
            if(self.head != None):
                self.head.pre = None
            return

        t = self.head
        for i in range(1, pos-1):
            t = t.next

        t.next = t.next.next
        if(t.next != None):
            t.next.pre = t

    def printDLL(self):
        t = self.head
        while(t != None):
            print(t.value, end="<-->")
            t = t.next
        print()
  
obj = DoublyLL()
obj.insertAtEnd(10)
obj.insertAtEnd(20)
obj.insertAtEnd(30)
obj.insertAtEnd(40)
obj.insertAtEnd(50)
obj.printDLL()
obj.insertAtBegin(5)
obj.insertAtPos(25, 4)


obj.printDLL()

        
```
