```python
class Stack:
    def __init__(self):
        self.s = []

    def length(self):
        return len(self.s)
    
    def push(self, data):
        self.s.insert(0, data)

    def peek(self):
        if self.s == []:
            raise Exception("Stack is Empty")
        else:
            return self.s[0]

    def pop(self):
        if self.s == []:
            raise Exception("Stack is Empty")
        else:
            return self.s.pop(0)

stack = Stack()

stack.push(10)
stack.push(20)
stack.push(30)

print(stack.length())
print(stack.pop())
print(stack.pop())
print(stack.pop())

```
