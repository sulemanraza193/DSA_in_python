```python
class Stack:
    def init__(self):
        self.items = []

    def length(self):
        return len(self.items)
    
    def push(self, value):
         self.items.insert(0, value)

    def Peek(self, value):
        if self.items == 0:
            raise Exception("Stack is Empty")
        else:
            return self.items[0]

    def pop(self):
        if self.items == 0:
            raise Exception("Stack is Empty")
        else:
            return self.items.pop(0)

stk = Stack()

stk.push(10)
stk.push(20)
stk.push(30)
print(stk)
```
